Imports

```python
from elasticsearch import Elasticsearch
from pprint import pprint
import csv
import sys
import pandas as pd
import json
```

Initialize Elasticsearch

```python
es = Elasticsearch("http://127.0.0.1:9200")
```

Create json that will be fed into elasticsearch

```python
df = pd.read_csv("IR2025/documents.csv", encoding='utf-8')

df["Title"] = df["Text"].apply(lambda x: x.split(":", 1)[0] if ":" in str(x) else "")
df["Body"] = df["Text"].apply(lambda x: x.split(":", 1)[1].strip() if ":" in str(x) else x)

records = df[["ID", "Title", "Body"]].to_dict(orient="records")

with open("IR2025/documents.jsonl", "w", encoding="utf-8") as f:
    for rec in records:
        f.write(json.dumps(rec, ensure_ascii=False) + "\n")
```
